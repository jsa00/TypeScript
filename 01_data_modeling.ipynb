## 문항 1 : 전자상거래 회원·상품 데이터 모델링

### 1. 인터페이스와 타입 별칭으로 데이터 구조 정의하기 [필수]
- 회원(User)과 상품(Product)이 어떤 모양의 데이터를 가져야 하는지 미리 규격을 정해둡니다.
- 정해진 값만 허용해야 하는 권한과 카테고리는 **문자열 리터럴 유니온 타입**을 타입 별칭으로 정의합니다.

#### 1.1 타입 별칭과 인터페이스 선택 기준

| **선언** | **방식** | **이유** |
| --- | --- | --- |
| `UserRole`, `Category` | 타입 별칭 (type) | 유니온 타입(`\|`)은 인터페이스로 정의할 수 없음 |
| `Identifiable` | 인터페이스 (interface) | 모든 엔티티가 공통으로 갖는 `id`를 한 곳에 정의하고 `extends`로 재사용 |
| `User`, `Product` | 인터페이스 (interface) | 객체 구조 정의 + `extends`로 확장하기 쉬움 |

선택적 속성(?)
> `discountRate?`처럼 속성명 뒤에 물음표를 붙이면 할인이 없는 상품은 해당 속성을 생략할 수 있습니다. 생략된 속성의 값은 `undefined`입니다.

#### 1.2 회원·상품 모델 정의 및 데이터 바인딩

```typescript
// 모든 엔티티의 공통 식별자
interface Identifiable {
    id: number; // 고유 식별 번호
}

// 회원 권한을 제한하는 리터럴 유니온 타입
type UserRole = "admin" | "customer" | "guest";

// Identifiable을 상속한 회원 모델
interface User extends Identifiable {
    name: string; // 회원 이름
    email: string; // 이메일 주소
    role: UserRole; // 회원 권한
    isActive: boolean; // 활성 여부
}

// 상품 카테고리를 제한하는 리터럴 유니온 타입
type Category = "electronics" | "clothing" | "books";

// Identifiable을 상속한 상품 모델
interface Product extends Identifiable {
    title: string; // 상품명
    price: number; // 정가
    category: Category; // 카테고리
    stock: number; // 재고 수량
    discountRate?: number; // 할인율(0~1)이며 생략할 수 있다.
}

// 회원 더미 데이터
const users: User[] = [
    { id: 1, name: "김관리", email: "admin@shop.com", role: "admin", isActive: true },
    { id: 2, name: "이고객", email: "lee@shop.com", role: "customer", isActive: true },
    { id: 3, name: "박고객", email: "park@shop.com", role: "customer", isActive: false },
    { id: 4, name: "최손님", email: "guest@shop.com", role: "guest", isActive: true },
];

// 상품 더미 데이터. discountRate는 생략할 수 있다.
const products: Product[] = [
    { id: 101, title: "무선 이어폰", price: 129000, category: "electronics", stock: 15, discountRate: 0.1 },
    { id: 102, title: "기계식 키보드", price: 89000, category: "electronics", stock: 7 },
    { id: 103, title: "오버핏 후드티", price: 45000, category: "clothing", stock: 30, discountRate: 0.2 },
    { id: 104, title: "타입스크립트 교과서", price: 32000, category: "books", stock: 12 },
];

console.log("회원 수:", users.length);
console.log("상품 수:", products.length);
```

### 2. 타입 규칙 위반 확인
- 타입 규칙에 어긋나는 데이터는 실행 전 **컴파일 단계에서 차단**됩니다.

| **위반 내용** | **예상 에러** |
| --- | --- |
| `UserRole`에 없는 값 `"manager"` | Type '"manager"' is not assignable to type 'UserRole' |
| `Category`에 없는 값 `"food"` | Type '"food"' is not assignable to type 'Category' |
| 필수 속성 `isActive` 누락 | Property 'isActive' is missing |
| `price`에 문자열 대입 | Type 'string' is not assignable to type 'number' |
